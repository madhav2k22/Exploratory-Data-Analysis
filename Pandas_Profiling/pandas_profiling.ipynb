# Pandas_Profiling
fg-data-profiling is the new, official name for the widely used ydata-profiling library (which itself was previously known as pandas-profiling).<br>
fg-data-profiling is an automated tool that generates a comprehensive, interactive report of your dataset's statistics, correlations, and data quality issues with just one line of code. It saves hours of manual work by instantly identifying missing values, outliers, and patterns in your data, whether you are using small Pandas dataframes or massive Spark clusters.


```python
import pandas as pd
from data_profiling import ProfileReport

titanic = pd.read_csv('../titanic.csv')
profile = ProfileReport(titanic, title='Titanic Dataset Profiling Report')
profile.to_file('titanic_report.html')
```
